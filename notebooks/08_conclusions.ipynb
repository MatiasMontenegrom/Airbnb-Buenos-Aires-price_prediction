## 2. Principales hallazgos

### Ingeniería y selección de variables

Se probaron variables derivadas como `beds_per_guest`, `bedrooms_per_guest`, `bathrooms_per_guest`, `beds_per_bedroom` y `amenities_count`. La mejora observada con feature engineering fue pequeña:

| Configuración | Variables | R² test | RMSE | MAE |
|---|---:|---:|---:|---:|
| Original | 136 | 0,5075 | 65.395,19 | 35.017,63 |
| Con feature engineering | 141 | 0,5105 | 65.197,79 | 34.976,25 |

PCA no mejoró los experimentos realizados con Ridge y SVR. Por eso, no se adoptó como estrategia principal.

### Modelo completo frente al compacto

| Métrica | HistGradientBoosting completo | HistGradientBoosting compacto |
|---|---:|---:|
| Variables | 136 | 43 |
| CV R² | 0,5221 | 0,5232 |
| CV RMSE | 60.612,38 | 60.544,91 |
| CV MAE | 34.555,90 | 34.548,93 |
| Test R² | 0,5075 | 0,5061 |
| Test RMSE | 65.395,19 | 65.488,10 |
| Test MAE | 35.017,63 | 35.158,32 |

La selección con umbral `0.5*mean` redujo las variables de 136 a 43 (aproximadamente un 68,4 % menos). La validación cruzada fue ligeramente mejor en el modelo compacto, pero sus métricas de test fueron un poco peores. Por lo tanto, no se puede afirmar que la selección aumente la precisión; sí simplifica el modelo con una diferencia pequeña en las métricas registradas.

## 3. Modelo recomendado

La elección depende de la prioridad:

- **Priorizar el rendimiento de test:** HistGradientBoosting completo, con 136 variables, obtuvo las métricas de test ligeramente mejores.
- **Priorizar simplicidad:** HistGradientBoosting compacto, con 43 variables, conservó un rendimiento muy parecido usando cerca de un tercio de las columnas.

Para presentar el proyecto, se recomienda reportar ambos y explicar el compromiso entre complejidad y rendimiento. No describir el modelo compacto como superior en precisión: los resultados de test no respaldan esa afirmación.

## 4. Cómo interpretar las métricas

- **R² ≈ 0,51 en test:** el modelo explica alrededor del 51 % de la variabilidad observada en los precios del conjunto de prueba. No significa que acierte el precio con un 51 % de precisión.
- **MAE ≈ 35.000:** el error absoluto medio es de aproximadamente 35.000 unidades de la variable objetivo.
- **RMSE ≈ 65.000:** el error cuadrático medio en raíz es mayor que el MAE, lo que sugiere que algunos errores grandes influyen de forma importante.

Los errores se expresan en la misma unidad y escala que el objetivo utilizado al entrenar el modelo. No deben interpretarse como dólares ni como otra moneda si el dataset no lo establece explícitamente.

## 5. Limitaciones metodológicas

1. **Reutilización del test:** el conjunto de test se consultó durante múltiples experimentos. Las métricas finales son útiles para comparar el flujo actual, pero no representan una evaluación completamente independiente.
2. **Selección de modelos:** los siguientes experimentos deberían basar sus decisiones en validación cruzada y reservar un conjunto de prueba nuevo o utilizar validación cruzada anidada para una estimación más imparcial.
3. **Calidad y representatividad de los datos:** los resultados dependen del período, la cobertura geográfica, los alojamientos incluidos y la calidad de los registros.
4. **Variables omitidas:** la predicción puede mejorar con información adicional fiable, por ejemplo fecha de la estadía, estacionalidad, eventos, políticas de cancelación u otras variables disponibles en el momento real de predecir.
5. **Interpretabilidad:** la importancia de una variable indica asociación predictiva dentro del modelo; no demuestra causalidad.
6. **Generalización:** el rendimiento en otros períodos o mercados puede diferir y debería comprobarse con datos externos o temporales.

## 7. Conclusión final

El proyecto permitió construir y comparar modelos para estimar el precio por noche de alojamientos de Airbnb en Buenos Aires. HistGradientBoosting fue uno de los candidatos con mejor rendimiento registrado, con un R² de test cercano a 0,51. El resultado principal no es que exista un modelo perfecto, sino que se realizó un proceso de experimentación: limpieza, ingeniería de variables, comparación de modelos, ajuste de hiperparámetros, reducción de características y evaluación de errores.
